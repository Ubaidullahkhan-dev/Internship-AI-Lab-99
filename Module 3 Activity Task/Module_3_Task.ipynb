# Activity 9 – Detect Income and Spending Outliers

```python
print("---Outliers Using Boxplot---")
import matplotlib.pyplot as plt
import seaborn as sns
num_col=df.select_dtypes(include=['int64','float64'])
plt.figure(figsize=(12,6))
sns.boxplot(data=num_col,orient='h')
plt.tight_layout()
plt.show()
```

```python
print("--- Counting Outliers using IQR Method ---")

outliers_count = {}

for col in num_col.columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
   
    total_outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    outliers_count[col] = total_outliers

for col, count in outliers_count.items():
    print(f"{col}: {count} outliers")
```

# Activity 10 – Handle Outliers

### For each identified outlier, decide whether to:

* **Remove**
* **Cap (Winsorization)**
* **Transform (Log Transformation)**
* **Keep (with justification)**

```python
print("--- Handel Outliers usnig Winsorization (Capping) Method---")
import numpy as np
for col in num_col.columns:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3-Q1

    lower = Q1-1.5*IQR

    upper = Q3+1.5*IQR

    df[col] = np.where(df[col]<lower, lower, df[col])

    df[col] = np.where(df[col]>upper, upper, df[col])
```

# Activity 13 – Create a Reusable Preprocessing Script

```python
def preprocess_data(path):

    df = pd.read_csv(path, sep="\t")

    df["Income"] = df["Income"].fillna(df["Income"].median())

    df = df.drop_duplicates()

    df["Dt_Customer"] = pd.to_datetime(
        df["Dt_Customer"],
        format="%d-%m-%Y",
        errors="coerce"
    )

    df["Education"] = df["Education"].str.strip().str.title()

    df["Marital_Status"] = (
        df["Marital_Status"]
        .str.strip()
        .str.title()
        .replace({
            "Together":"Married",
            "Absurd":"Single",
            "Yolo":"Single"
        })
    )

    current_year = pd.Timestamp.now().year

    df["Age"] = current_year - df["Year_Birth"]

    df = df[
        (df["Age"]<=100) &
        (df["Age"]>=18)
    ]

    return df
```

```python
clean_df = preprocess_data("customer data/marketing_campaign.csv")

clean_df.head()
```
