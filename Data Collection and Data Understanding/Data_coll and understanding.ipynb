# Task 5 – Identify Missing Values


```python
print("---Missing Values---")
missing_values = df.isnull().sum().sort_values(ascending=False)
print(missing_values)
```

### Missing Value Percentage

```python
print("---Missing Value Percentage---")
missing_percentage=(missing_values/len(df))*100
print(missing_percentage)
```

### Create a summary table

```python
print("---Create a summary table---")
missing_report = pd.DataFrame({
    "Variable": missing_values.index,
    "Missing Values": missing_values.values,
    "Percentage (%)": missing_percentage.values
})
print(missing_report)
```

```python
print("---Columns Only Have Missing Values---")
print(missing_report[missing_report["Missing Values"] > 0])
```

# Task 7 – Identify Incorrect or Inconsistent Values

### Invalid Age

### Negaive income


```python
negative_income=df[(df['Income']<0)]
print(negative_income)
```

### Empty Categorie

```python
empty_education=df[df['Education'].astype(str).str.strip()==""]
empty_marital=df[df['Marital_Status'].astype(str).str.strip()==""]

print("--- Checking for Empty Spaces in Education Columns ---")

if len(empty_education)>0:
    print(f"Education: Yes ({len(empty_education)} empty spaces found)")
else:
    print("No Empty Columns ")


print("--- Checking for Empty Spaces in Material Columns ---")

if len(empty_marital)<0:
    print(f"Marital_Status: Yes ({len(empty_marital)} empty spaces found)")
else:
    print("No Empty Columns ")
```

### Incorrect dates

```python
df['Dt_Customer']=pd.to_datetime(
    df['Dt_Customer'],
    format="%d-%m-%Y",
    errors='coerce'
)
invalid_dates=df[df["Dt_Customer"].isnull()]
if len(invalid_dates)>0:
    print(f"Yes Date Column have Inavlid Dates in {len(invalid_dates)}")
else:
    print("No invalid Data")
```
