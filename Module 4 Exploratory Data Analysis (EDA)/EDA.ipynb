# Task 6: Customer Recency Analysis

```python



# 1. DISTRIBUTION OF RECENCY (Histplot + Density Plot)

plt.figure(figsize=(10, 5))
sns.histplot(df['Recency'], kde=True, color='purple', bins=20)
plt.title('Distribution of Customer Recency')
plt.xlabel('Days Since Last Purchase (Recency)')
plt.ylabel('Count of Customers')
plt.show()



# 2. SEPARATING ACTIVE & INACTIVE CUSTOMERS (Creating a Status Column)

# Standard rule: 30 din ya usse kam wale Active, 60 din se zyada wale Inactive
def check_status(days):
    if days <= 30:
        return 'Active'
    elif days <= 60:
        return 'Moderate'
    else:
        return 'Inactive (Requires Re-engagement)'

df['Customer_Status'] = df['Recency'].apply(check_status)



# 3. COUNT PLOT (Visualizing Active vs Inactive Segments)

plt.figure(figsize=(8, 5))
sns.countplot(
    x='Customer_Status', 
    data=df, 
    hue='Customer_Status',
    palette='viridis',
    legend=False,
    order=['Active', 'Moderate', 'Inactive (Requires Re-engagement)']
)
plt.title('Customer Activity Status Breakdown')
plt.xlabel('Activity Classification')
plt.ylabel('Number of Customers')
plt.show()



# 4. BUSINESS INSIGHTS REPORT (Professional English)

print("\n==============================================")
print("       CUSTOMER RECENCY & ACTIVITY REPORT     ")
print("==============================================")

total_cust = len(df)
active_count = df[df['Customer_Status'] == 'Active'].shape[0]
inactive_count = df[df['Customer_Status'] == 'Inactive (Requires Re-engagement)'].shape[0]

active_pct = (active_count / total_cust) * 100
inactive_pct = (inactive_count / total_cust) * 100

print(f"• Recency Distribution: The days since last purchase range uniformly from {df['Recency'].min()} to {df['Recency'].max()} days.")
print(f"• Active Customer Base: {active_count} customers ({active_pct:.1f}%) are highly active, having made a purchase within the last 30 days.")
print(f"• Re-engagement Target: {inactive_count} customers ({inactive_pct:.1f}%) have been inactive for over 60 days.")
print("• Strategic Recommendation: Targeted email or re-engagement campaigns should immediately focus on the Inactive segment to recover churned users.")
print("==============================================")
```

# Task 7: Marketing Campaign Analysis

```python


# Columns aur unke labels
camp_cols = ['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5', 'Response']
camp_labels = ['Campaign 1', 'Campaign 2', 'Campaign 3', 'Campaign 4', 'Campaign 5', 'Latest Campaign']

# 1. DATA CLEANING & TYPE CONVERSION
for col in camp_cols:
    if col in df.columns:
        # Object/Text data ko numeric me convert karna aur NaN ko 0 banana
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Calculations (Ab bilkul sahi numbers milenge)
total_accepted = df[camp_cols].sum()
# Zero division se bachne ke liye safe calculation
total_customers = len(df) if len(df) > 0 else 1
acceptance_rates = (total_accepted / total_customers) * 100


# =====================================================================
# 2. BAR CHART
# =====================================================================
plt.figure(figsize=(10, 5))
sns.barplot(
    x=camp_labels, 
    y=acceptance_rates.values, 
    hue=camp_labels,
    palette='muted',
    legend=False
)
plt.title('Campaign Success Rates Comparison (Bar Chart)')
plt.xlabel('Marketing Campaigns')
plt.ylabel('Acceptance / Response Rate (%)')
plt.show()


# =====================================================================
# 3. SAFE PIE CHART (Ab crash nahi hoga)
# =====================================================================
plt.figure(figsize=(7, 7))
# Agar data me saare zeros hain toh daira nahi banega, text show hoga
if total_accepted.sum() == 0:
    plt.text(0.5, 0.5, 'No Campaigns Accepted (All Sums are 0.0)\nPie Chart Cannot Be Rendered', 
             ha='center', va='center', fontsize=12, color='red', weight='bold')
    plt.title('Contribution Share of Each Campaign (No Data Available)')
    plt.axis('off')
else:
    plt.pie(
        total_accepted.values, 
        labels=camp_labels, 
        autopct='%1.1f%%', 
        colors=plt.get_cmap('Pastel1').colors
    )
    plt.title('Contribution Share of Each Campaign (Pie Chart)')
plt.show()


# =====================================================================
# 4. COUNT PLOT
# =====================================================================
df['Total_Campaigns_Accepted'] = df[camp_cols].sum(axis=1)

plt.figure(figsize=(8, 5))
sns.countplot(
    x='Total_Campaigns_Accepted', 
    data=df, 
    hue='Total_Campaigns_Accepted',
    palette='Blues',
    legend=False
)
plt.title('Number of Campaigns Accepted per Customer (Count Plot)')
plt.xlabel('Total Campaigns Accepted')
plt.ylabel('Number of Customers')
plt.show()


# =====================================================================
# 5. BUSINESS INSIGHTS REPORT (Professional English)
# =====================================================================
print("\n==============================================")
print("       MARKETING CAMPAIGN PERFORMANCE REPORT   ")
print("==============================================")

most_successful_idx = total_accepted.values.argmax()
most_successful_name = camp_labels[most_successful_idx]
highest_rate = acceptance_rates.values.max()
latest_camp_rate = acceptance_rates.get('Response', 0)

# Mode ko safely nikalna bina float crash ke
mode_values = df['Total_Campaigns_Accepted'].mode()
frequent_mode = int(mode_values.iloc[0]) if not mode_values.empty and not pd.isna(mode_values.iloc[0]) else 0

print(f"• Top Performing Campaign: {most_successful_name} currently reflects the baseline metric with {highest_rate:.1f}% acceptance.")
print(f"• Latest Campaign Performance: The final marketing offer recorded an evaluation rate of {latest_camp_rate:.1f}%.")
print(f"• Customer Conversion: The data indicates that the predominant segment of customers accepted {frequent_mode} campaigns.")
print("• Strategic Takeaway: Since the current dataset contains baseline zero responses for active campaigns, data enrichment is recommended before scaling future budgets.")
print("==============================================")
```

# Task 9: Correlation Analysis

```python

# 1. Main numerical columns ki list jo humne pichle tasks me banayi/use ki hain
# Agar koi column aapke dataset me missing ho toh wo auto-ignore ho jayega
core_numerical_cols = ['Age', 'Income', 'Total_Spending', 'Total_Num_Purchases', 'Recency', 'NumWebVisitsMonth']
valid_cols = [col for col in core_numerical_cols if col in df.columns]

# Ensure data is completely numeric and clean
for col in valid_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# =====================================================================
# 2. PEARSON CORRELATION MATRIX CALCULATE KARNA
# =====================================================================
corr_matrix = df[valid_cols].corr(method='pearson')


# =====================================================================
# 3. HEATMAP PLOT
# =====================================================================
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix, 
    annot=True,             # Squares ke andar numbers likhna
    cmap='coolwarm',        # Red (Positive), Blue (Negative), Light (Neutral)
    fmt='.2f',              # Decimal ke baad 2 numbers
    linewidths=0.5,
    vmin=-1, vmax=1         # Scale ko barabar -1 se 1 tak rakhna
)
plt.title('Pearson Correlation Matrix Heatmap', fontsize=14, weight='bold')
plt.show()


# =====================================================================
# 4. DYNAMIC BUSINESS INSIGHTS REPORT (Professional English)
# =====================================================================
print("\n==============================================")
print("             CORRELATION ANALYSIS REPORT       ")
print("==============================================")

# Automatically finding relationships from the calculated matrix
pos_pairs = []
neg_pairs = []
ind_pairs = []

# Loop over the matrix to identify the requested relationship types
cols = corr_matrix.columns
for i in range(len(cols)):
    for j in range(i+1, len(cols)):
        val = corr_matrix.iloc[i, j]
        pair_str = f"{cols[i]} and {cols[j]} ({val:.2f})"
        
        if val >= 0.4:
            pos_pairs.append(pair_str)
        elif val <= -0.4:
            neg_pairs.append(pair_str)
        elif abs(val) <= 0.1:
            ind_pairs.append(pair_str)

# Formatting output safely
strong_pos = ", ".join(pos_pairs) if pos_pairs else "None identified above threshold"
strong_neg = ", ".join(neg_pairs) if neg_pairs else "None identified above threshold"
independent_vars = ", ".join(ind_pairs[:4]) if ind_pairs else "None identified near zero"

print(f"• Strong Positive Correlations (>= 0.4): \n  {strong_pos}\n")
print(f"• Strong Negative Correlations (<= -0.4): \n  {strong_neg}\n")
print(f"• Independent Variables (Near 0.0): \n  {independent_vars}\n")
print("• Strategic Conclusion: Data drivers with high positive correlation indicate that target marketing should bundle these features, while independent features represent completely distinct customer segments.")
print("==============================================")
```

# Task 10: Customer Segmentation

```python


# Ensure all used columns are clean and numeric
check_cols = ['Total_Spending', 'Total_Num_Purchases', 'NumDealsPurchases', 'Total_Campaigns_Accepted', 'Recency']
for col in check_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# 1. SEGMENTATION LOGIC (Percentiles nikalte hain taaki data dynamically fit ho)
spend_75 = df['Total_Spending'].quantile(0.75)
purch_75 = df['Total_Num_Purchases'].quantile(0.75)
deals_75 = df['NumDealsPurchases'].quantile(0.75)

def assign_segment(row):
    # High-Value: Jo bohot zyada spending kar rahe hain
    if row['Total_Spending'] >= spend_75:
        return 'High-Value Customer'
    # Inactive: Jo bohot din se nahi aaye (Recency > 60 days)
    elif row['Recency'] > 60:
        return 'Inactive Customer'
    # Frequent Buyer: Jo bohot baar kharidari karte hain
    elif row['Total_Num_Purchases'] >= purch_75:
        return 'Frequent Buyer'
    # Discount Seeker: Jo bargains/deals zyada lete hain
    elif row['NumDealsPurchases'] >= deals_75:
        return 'Discount Seeker'
    # Campaign Responder: Jinhone marketing offers accept ki hon
    elif row['Total_Campaigns_Accepted'] > 0:
        return 'Campaign Responder'
    # Baki sab standard users
    else:
        return 'Standard/Low-Value Customer'

# Segment column banayein
df['Customer_Segment'] = df.apply(assign_segment, axis=1)


# =====================================================================
# 2. VISUALIZATION (Count Plot of Customer Segments)
# =====================================================================
plt.figure(figsize=(12, 6))
sns.countplot(
    y='Customer_Segment', 
    data=df, 
    hue='Customer_Segment',
    palette='Set2',
    legend=False,
    order=df['Customer_Segment'].value_counts().index
)
plt.title('Customer Segmentation Breakdown', fontsize=14, weight='bold')
plt.xlabel('Number of Customers')
plt.ylabel('Identified Segments')
plt.tight_layout()
plt.show()


# =====================================================================
# 3. PROFESSIONAL BUSINESS INSIGHTS REPORT (English)
# =====================================================================
print("\n==============================================")
print("             CUSTOMER SEGMENTATION REPORT      ")
print("==============================================")

segment_counts = df['Customer_Segment'].value_counts()
total_records = len(df)

for segment, count in segment_counts.items():
    pct = (count / total_records) * 100
    print(f"• {segment}: {count} customers ({pct:.1f}%)")

print("\n• Strategic Marketing Takeaways:")
print("  1. High-Value Segments should be targeted with loyalty programs and premium product previews.")
print("  2. Inactive Customers require immediate re-engagement discount triggers via email channels.")
print("  3. Discount Seekers should be funneled towards clearance and bundle promotion campaigns.")
print("==============================================")
```
