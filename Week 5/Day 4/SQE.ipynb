```python
BASE_PATH = Path(r"D:\AI-Lab-99")

WEEK5_PATH = BASE_PATH / "Week 5"

DATA_PATH = WEEK5_PATH / "Day 1" / "Final Dataset.csv"
MODEL_PATH = WEEK5_PATH / "Day 4" / "kmeans_model.pkl"
RESULTS_PATH = WEEK5_PATH / "Day 4" / "day3_results.pkl"

print("DATA:", DATA_PATH)
print("MODEL:", MODEL_PATH)
print("RESULTS:", RESULTS_PATH)
```

```python
def find_file(filename, base_path):
    for root, dirs, files in os.walk(base_path):
        if filename in files:
            return Path(root) / filename
    return None


csv_path = find_file("Final Dataset.csv", BASE_PATH)
model_path = find_file("kmeans_model.pkl", BASE_PATH)
results_path = find_file("day3_results.pkl", BASE_PATH)

print(" FILE DETECTION")
print("=" * 60)

print("Final Dataset.csv :", csv_path)
print("kmeans_model.pkl  :", model_path)
print("day3_results.pkl  :", results_path)
```

```python

# CELL 15 — CREATE QA RESULTS


import pandas as pd

# QA test results
qa_results = []

def add_test(test_name, condition, details=""):
    qa_results.append({
        "Test": test_name,
        "Status": "PASS" if condition else "FAIL",
        "Details": details
    })


# 1. Dataset exists
add_test(
    "Dataset Availability",
    "df" in globals() and isinstance(df, pd.DataFrame),
    "Main dataframe checked"
)


# 2. Dataset not empty
if "df" in globals() and isinstance(df, pd.DataFrame):
    add_test(
        "Dataset Not Empty",
        len(df) > 0,
        f"Rows: {len(df)}"
    )


# 3. Duplicate check
if "df" in globals() and isinstance(df, pd.DataFrame):
    duplicates = df.duplicated().sum()

    add_test(
        "Duplicate Check",
        duplicates == 0,
        f"Duplicates: {duplicates}"
    )


# 4. Missing values
if "df" in globals() and isinstance(df, pd.DataFrame):
    missing = df.isnull().sum().sum()

    add_test(
        "Missing Values Check",
        missing == 0,
        f"Missing values: {missing}"
    )


# 5. Cluster column
if "df" in globals() and isinstance(df, pd.DataFrame):
    cluster_exists = "Cluster" in df.columns

    add_test(
        "Cluster Column",
        cluster_exists,
        "Cluster column available"
    )


# 6. Model file
from pathlib import Path

model_paths = list(
    Path(r"D:\AI-Lab-99").rglob("kmeans_model.pkl")
)

add_test(
    "KMeans Model File",
    len(model_paths) > 0,
    str(model_paths[0]) if model_paths else "Model not found"
)


# Create dataframe
results_test_df = pd.DataFrame(qa_results)

# Also create compatible variable
qa_results_df = results_test_df.copy()

print(" QA RESULTS CREATED")
print("=" * 60)

display(results_test_df)
```

```python

# CELL 16 — FINAL QA SUMMARY


total_tests = len(results_test_df)

passed_tests = (
    results_test_df["Status"] == "PASS"
).sum()

failed_tests = (
    results_test_df["Status"] == "FAIL"
).sum()

print(" FINAL QA SUMMARY")
print("=" * 60)

print("Total Tests :", total_tests)
print("Passed      :", passed_tests)
print("Failed      :", failed_tests)

if failed_tests == 0:
    print("\n ALL QA TESTS PASSED")
else:
    print("\n SOME QA TESTS FAILED — REVIEW REQUIRED")
```
