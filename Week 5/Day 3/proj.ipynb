```python
import os
import joblib
import pandas as pd
import numpy as np
```

```python

# TASK 1 — FIND KMEANS MODEL


base_path = r"D:\AI-Lab-99"

model_path = None

for root, dirs, files in os.walk(base_path):
    for file in files:
        
        if file.lower() == "kmeans_model.pkl":
            model_path = os.path.join(root, file)
            break
    
    if model_path is not None:
        break


if model_path is not None:
    print(" K-Means model found!")
    print("Model path:")
    print(model_path)
else:
    print(" kmeans_model.pkl not found.")
```

```python

# LOAD SAVED KMEANS MODEL


kmeans_model = joblib.load(model_path)

print(" K-Means model loaded successfully.")
print("Model type:", type(kmeans_model).__name__)
```

```python

# MODEL INFORMATION


print("Model Type:", type(kmeans_model).__name__)

if hasattr(kmeans_model, "n_clusters"):
    print("Number of Clusters:", kmeans_model.n_clusters)

if hasattr(kmeans_model, "n_features_in_"):
    print("Number of Input Features:", kmeans_model.n_features_in_)

if hasattr(kmeans_model, "cluster_centers_"):
    print("Cluster Centers Shape:", kmeans_model.cluster_centers_.shape)
```

```python

# TASK 2 — FIND PREPROCESSING / SCALER FILES


preprocessing_files = []

for root, dirs, files in os.walk(base_path):
    for file in files:

        file_lower = file.lower()

        if (
            file_lower.endswith(".pkl")
            or file_lower.endswith(".joblib")
        ):

            if file_lower != "kmeans_model.pkl":
                preprocessing_files.append(
                    os.path.join(root, file)
                )


print("Possible preprocessing/model files:")

if preprocessing_files:

    for path in preprocessing_files:
        print("", path)

else:
    print(" No additional .pkl/.joblib files found.")
```

## TASK 8 — CUSTOMER SEGMENT PREDICTION

```python
def predict_customer_segment(customer_data):

    try:
        # Make sure columns are in exact model order
        customer_data = customer_data[model_features]

        # Predict cluster
        cluster_prediction = kmeans_model.predict(
            customer_data
        )[0]

        return int(cluster_prediction)

    except Exception as e:

        print(" Prediction Error:")
        print(e)

        return None
```

```python

# PREDICTION RESULT


sample_customer = X_model.iloc[[0]].copy()

predicted_cluster = predict_customer_segment(
    sample_customer
)

print("Predicted Cluster:", predicted_cluster)
```

```python

# CLUSTER → BUSINESS SEGMENT


if predicted_cluster is not None:

    if "Cluster" in df.columns and "Customer_Segment" in df.columns:

        mapping = (
            df[["Cluster", "Customer_Segment"]]
            .drop_duplicates()
        )

        match = mapping[
            mapping["Cluster"] == predicted_cluster
        ]

        if not match.empty:

            business_segment = match[
                "Customer_Segment"
            ].iloc[0]

            print("Predicted Cluster:", predicted_cluster)
            print(
                "Business Segment:",
                business_segment
            )

        else:

            business_segment = "Unknown Segment"

            print(
                " Business segment not found."
            )

    else:

        business_segment = f"Cluster {predicted_cluster}"

        print(
            "Predicted Cluster:",
            predicted_cluster
        )
```

```python

#  MARKETING RECOMMENDATION LOGIC


# 1. COMMUNICATION CHANNEL


if "Preferred_Shopping_Channel" in segment_customers.columns:

    channel_counts = (
        segment_customers["Preferred_Shopping_Channel"]
        .value_counts()
    )

    preferred_channel = channel_counts.index[0]

else:

    preferred_channel = "Email"


# 2. DISCOUNT / OFFER STRATEGY

if "Deal_Dependency" in segment_customers.columns:

    deal_values = (
        segment_customers["Deal_Dependency"]
        .astype(str)
        .str.lower()
    )

    if deal_values.str.contains("high").mean() > 0.5:

        discount_strategy = "Special Discount / Coupon"
        personalized_offer = "Limited-Time Discount Offer"

    else:

        discount_strategy = "Low Discount / Value-Based Offer"
        personalized_offer = "Exclusive Product Offer"

else:

    discount_strategy = "Personalized Offer"
    personalized_offer = "Special Customer Offer"



# 3. MARKETING MESSAGE


if risk_level in ["High", "Medium"]:

    marketing_message = (
        "We Miss You! Come Back and Enjoy "
        "a Special Personalized Offer"
    )

    campaign_timing = "Immediate Win-Back Campaign"

elif "Total_spending" in segment_customers.columns:

    segment_spending = (
        segment_customers["Total_spending"].mean()
    )

    overall_spending = df["Total_spending"].mean()

    if segment_spending > overall_spending:

        marketing_message = (
            "Enjoy Exclusive Rewards and "
            "Premium Products Selected for You"
        )

        campaign_timing = "Weekly"

    else:

        marketing_message = (
            "Discover Great Products and "
            "Save More with Personalized Offers"
        )

        campaign_timing = "Bi-Weekly"

else:

    marketing_message = (
        "Discover Products Selected Especially for You"
    )

    campaign_timing = "Bi-Weekly"



# 4. PRODUCT RECOMMENDATION


if "preferred_product" in globals():

    primary_product = preferred_product

else:

    primary_product = "Personalized Product Category"



# 5. RETENTION STRATEGY


if risk_level == "High":

    retention_strategy = (
        "Priority retention campaign with "
        "win-back offer"
    )

elif risk_level == "Medium":

    retention_strategy = (
        "Regular engagement and personalized offers"
    )

else:

    retention_strategy = (
        "Loyalty rewards and exclusive benefits"
    )


print(" Marketing recommendations generated.")
```
