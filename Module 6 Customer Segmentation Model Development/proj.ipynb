# Task 5 — Determine Optimal Number of Clusters

```python
from sklearn.cluster import KMeans
wcss = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(df)

    wcss.append(kmeans.inertia_)
```

## Plot the Curve Elbow

```python
plt.figure(figsize=(8,5))

plt.plot(range(2,11) ,wcss , maker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Inertia)")

plt.grid(True)

plt.show()
```

## Display WCSS Values

```python
for k, inertia in zip(range(2,11), wcss):
    print(f"K = {k}  -->  WCSS = {inertia:.2f}")
```

# Task 6 — Silhouette Score Analysis

```python
from sklearn.metrics import silhouette_score, silhouette_samples

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(df)

    score = silhouette_score(df, labels)

    silhouette_scores.append(score)

    print(f"K = {k} --> Silhouette Score = {score:.4f}")
```

### Plot Silhouette Scores

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for Different K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.grid(True)

plt.show()
```

### Best K

```python
import numpy as np
best_k = np.argmax(silhouette_scores) + 2

print("Best K =", best_k)
print("Best Silhouette Score =", max(silhouette_scores))
```

### Silhouette Plot (Best K)

```python
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(df)

sample_scores = silhouette_samples(df, labels)

plt.figure(figsize=(8,6))

y_lower = 10

for i in range(best_k):

    cluster_scores = sample_scores[labels == i]
    cluster_scores.sort()

    size = cluster_scores.shape[0]
    y_upper = y_lower + size

    plt.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        cluster_scores,
        alpha=0.7
    )

    plt.text(-0.05, y_lower + size/2, str(i))

    y_lower = y_upper + 10

plt.axvline(
    x=max(silhouette_scores),
    color="red",
    linestyle="--"
)

plt.title("Silhouette Plot")
plt.xlabel("Silhouette Coefficient")
plt.ylabel("Cluster")

plt.show()
```

# Task 7 — Davies–Bouldin Index

```python
from sklearn.metrics import davies_bouldin_score
```

## Calculate DB Index (K = 2 to 10)

```python
db_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(df)

    score = davies_bouldin_score(df, labels)

    db_scores.append(score)

    print(f"K = {k} --> DB Index = {score:.4f}")
```

## Plot DB Index

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    db_scores,
    marker="o"
)

plt.title("Davies-Bouldin Index")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("DB Index")

plt.grid(True)

plt.show()
```

## Best K

```python
best_db_k = db_scores.index(min(db_scores)) + 2

print("Best K =", best_db_k)
print("Lowest DB Index =", min(db_scores))
```

## Comparison Table

# Task 11 — DBSCAN Clustering

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df_dbscan = df.copy()

df_dbscan["Cluster"] = dbscan_labels

df_dbscan.head()
```

```python
n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)

print("Number of Clusters:", n_clusters)
```

### Noise Points (Outliers)

```python
noise_points = np.sum(dbscan_labels == -1)

print("Noise Points:", noise_points)
```

### Cluster Size

```python
cluster_size = pd.Series(dbscan_labels).value_counts().sort_index()

print(cluster_size)
```

### Silhouette Score

```python
if n_clusters >= 2:

    mask = dbscan_labels != -1

    dbscan_silhouette = silhouette_score(
        df[mask],
        dbscan_labels[mask]
    )

    print("Silhouette Score:", round(dbscan_silhouette,4))

else:
    dbscan_silhouette = None
    print("Silhouette Score cannot be calculated.")
```

# Task 15 — Cluster Stability Analysis

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.utils import resample
import pandas as pd
```

### Different Random States

```python
random_states = [0, 10, 20, 42, 100]

results = []

for state in random_states:

    model = KMeans(
        n_clusters=best_k,
        random_state=state,
        n_init=10
    )

    labels = model.fit_predict(df)

    score = silhouette_score(df, labels)

    results.append([state, round(score,4)])

random_state_df = pd.DataFrame(
    results,
    columns=["Random State", "Silhouette Score"]
)

random_state_df
```

### Different Train Samples

```python
sample_results = []

for seed in [1, 10, 20, 42, 100]:

    sample = df.sample(
        frac=0.8,
        random_state=seed
    )

    model = KMeans(
        n_clusters=best_k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(sample)

    score = silhouette_score(sample, labels)

    sample_results.append([seed, round(score,4)])

sample_df = pd.DataFrame(
    sample_results,
    columns=["Sample Seed", "Silhouette Score"]
)

sample_df
```

### Bootstrap Sampling

```python
bootstrap_results = []

for i in range(5):

    bootstrap_sample = resample(
        df,
        replace=True,
        random_state=i
    )

    model = KMeans(
        n_clusters=best_k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(bootstrap_sample)

    score = silhouette_score(
        bootstrap_sample,
        labels
    )

    bootstrap_results.append([i+1, round(score,4)])

bootstrap_df = pd.DataFrame(
    bootstrap_results,
    columns=["Bootstrap Sample", "Silhouette Score"]
)

bootstrap_df
```

### Stability Summary

```python
print("Random State Results")
display(random_state_df)

print("Train Sample Results")
display(sample_df)

print("Bootstrap Results")
display(bootstrap_df)
```

### Save Stability Report

```python
random_state_df.to_csv(
    "Random_State_Stability.csv",
    index=False
)

sample_df.to_csv(
    "Sample_Stability.csv",
    index=False
)

bootstrap_df.to_csv(
    "Bootstrap_Stability.csv",
    index=False
)

print("Stability reports saved successfully.")
```
